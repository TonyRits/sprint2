# Statistics in Machine Learning

Statistics helps us prepare data, understand patterns, build models, and evaluate Machine Learning results.

This notebook covers important statistical and Machine Learning concepts used in practical ML workflows.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler, MinMaxScaler
from sklearn.linear_model import LinearRegression, LogisticRegression
from sklearn.naive_bayes import GaussianNB
from sklearn.cluster import KMeans
from sklearn.decomposition import PCA
from sklearn.metrics import (
    mean_squared_error,
    accuracy_score,
    classification_report
)
```

## 1. Data Cleaning

### Definition
Data cleaning is the process of identifying and correcting incorrect, duplicate, inconsistent, or invalid data.

### Purpose
Clean data improves the quality and reliability of Machine Learning models.

Use data cleaning before training a Machine Learning model.

A company may have duplicate customer records, missing ages, and incorrect product prices.

These issues should be cleaned before building a customer prediction model.

```python
data = pd.DataFrame({
    "Age": [25, 30, np.nan, 40],
    "Salary": [30000, 40000, 50000, 60000],
    "City": ["Chennai", "Chennai", "Mumbai", "Chennai"]
})

print("Missing values:")
print(data.isnull().sum())

data["Age"] = data["Age"].fillna(data["Age"].median())

print("\nCleaned Data:")
print(data)
```

### Interpretation

The missing Age value was replaced with the median age.

This makes the dataset more suitable for Machine Learning.

## 2. Missing Value Handling

### Definition
Missing value handling means identifying and treating empty or unavailable values in a dataset.

### Common methods
- Mean
- Median
- Mode
- Forward fill
- Backward fill
- Removing rows or columns

Use missing value handling when datasets contain NaN or empty values.

If some customers have missing income values, the missing values can be replaced using an appropriate statistical method.

```python
data = pd.DataFrame({
    "Age": [20, 25, np.nan, 35, 40],
    "Salary": [20000, 25000, 30000, np.nan, 40000]
})

data["Age"] = data["Age"].fillna(data["Age"].median())
data["Salary"] = data["Salary"].fillna(data["Salary"].mean())

print(data)
```

### Interpretation

Median and mean are commonly used to replace missing numerical values.

The appropriate method depends on the data distribution and business context.

## 3. Outlier Detection

### Definition
An outlier is an observation that is unusually different from most other observations.

### Common methods
- IQR method
- Z-score method
- Box plot

Use outlier detection when extreme values may affect analysis or model performance.

A dataset contains customer purchases of ₹500, ₹700, ₹800, and ₹100000.

The ₹100000 value may need investigation as a possible outlier.

```python
data = np.array([10, 12, 11, 13, 12, 50])

Q1 = np.percentile(data, 25)
Q3 = np.percentile(data, 75)

IQR = Q3 - Q1

lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR

outliers = data[(data < lower) | (data > upper)]

print("Outliers:", outliers)
```

## 4. Feature Engineering

### Definition
Feature engineering creates or transforms input variables to provide useful information to a Machine Learning model.

### Purpose
Good features can help a model learn patterns more effectively.

Instead of using Date directly, we can create Year, Month, Day, and Day_of_Week features for a sales prediction model.

```python
data = pd.DataFrame({
    "Price": [100, 200, 300],
    "Quantity": [2, 3, 4]
})

data["Total_Sales"] = data["Price"] * data["Quantity"]

print(data)
```

### Interpretation

`Total_Sales` is a new feature created from Price and Quantity.

This is an example of feature engineering.

## 5. Feature Scaling

### Definition
Feature scaling brings numerical features to comparable scales.

Two common methods are Standardization and Normalization.

Scaling is important for algorithms that are sensitive to feature magnitude, such as KNN, SVM, K-Means, and Logistic Regression.

If Age ranges from 20–60 but Salary ranges from 20,000–100,000, scaling prevents Salary from dominating because of its larger numerical range.    

```python
data = np.array([
    [20, 20000],
    [30, 40000],
    [40, 60000],
    [50, 80000]
])

scaler = StandardScaler()

scaled_data = scaler.fit_transform(data)

print(scaled_data)
```

### Interpretation

StandardScaler transforms features so they have approximately mean 0 and standard deviation 1.

## 6. Data Normalization

### Definition
Normalization usually scales values to a fixed range, commonly 0 to 1.

Formula:

Normalized value = (x - minimum) / (maximum - minimum)

Use normalization when features have very different ranges and the algorithm benefits from bounded values.

Customer age and spending amount can have very different ranges.

Min-Max normalization can convert both features to a common 0–1 range.

```python
data = np.array([
    [20, 20000],
    [30, 40000],
    [40, 60000],
    [50, 80000]
])

scaler = MinMaxScaler()

normalized_data = scaler.fit_transform(data)

print(normalized_data)
```

### Interpretation

The feature values are transformed into a range between 0 and 1.

Normalization can be useful for distance-based algorithms.

## 7. Model Evaluation

### Definition
Model evaluation measures how well a Machine Learning model performs on data.

Different metrics are used for regression and classification.

Regression:
- MAE
- MSE
- RMSE
- R²

Classification:
- Accuracy
- Precision
- Recall
- F1-score

A company can evaluate a sales prediction model using MAE or RMSE to understand prediction errors.

```python
actual = [100, 200, 300, 400]
predicted = [110, 190, 310, 380]

mse = mean_squared_error(actual, predicted)

print("Mean Squared Error:", mse)
```

### Interpretation

Lower MSE generally indicates smaller prediction errors.

The appropriate evaluation metric depends on the Machine Learning problem.

## 8. Linear Regression

### Definition
Linear Regression predicts a continuous target using one or more input features.

Example: predicting sales from advertising expenditure.

A company can predict sales based on advertising expenditure.

```python
X = np.array([[10], [20], [30], [40], [50]])
y = np.array([15, 25, 35, 45, 55])

model = LinearRegression()

model.fit(X, y)

prediction = model.predict([[60]])

print("Predicted Sales:", prediction[0])
```

### Interpretation

The model learns a linear relationship between advertising expenditure and sales.

It can then predict sales for a new advertising value.

## 9. Logistic Regression

### Definition
Logistic Regression is a classification algorithm used to predict class probabilities.

It is commonly used for binary classification.

A company can predict whether a customer will churn or not churn.

```python
X = np.array([[20],[25],[30],[35],[40],[45]])

y = np.array([0, 0, 0, 1, 1, 1])

model = LogisticRegression()

model.fit(X, y)

prediction = model.predict([[38]])
probability = model.predict_proba([[38]])

print("Prediction:", prediction[0])
print("Probability:", probability)
```

### Interpretation

The model predicts a class and provides the probability of each class.

For binary classification, the output can represent classes such as 0/1 or No/Yes.

## 10. Naive Bayes

### Definition
Naive Bayes is a probabilistic classification algorithm based on Bayes' theorem.

It assumes that features are conditionally independent given the class.

Naive Bayes can be used for spam email classification or sentiment classification.

```python
X = np.array([[1, 0],[1, 1],[0, 1],[0, 0],[1, 0],[0, 1]])

y = np.array([1, 1, 0, 0, 1, 0])

model = GaussianNB()

model.fit(X, y)

prediction = model.predict([[1, 1]])

print("Prediction:", prediction[0])
```

### Interpretation

Naive Bayes uses probability to determine the most likely class for a new observation.

It is commonly useful for text and classification problems.

## 11. Clustering

### Definition
Clustering is an unsupervised learning technique that groups similar observations together.

K-Means is a common clustering algorithm.

A company can group customers into different segments based on spending and purchase behavior.

```python
X = np.array([[10, 20],[12, 22],[50, 60],[52, 62],[90, 100],[92, 102]])

model = KMeans(n_clusters=3, random_state=42, n_init=10)

labels = model.fit_predict(X)

print("Cluster Labels:", labels)
print("Cluster Centers:")
print(model.cluster_centers_)
```

### Interpretation

K-Means assigns each observation to a cluster based on similarity.

Observations within the same cluster are relatively similar to each other.

## 12. Principal Component Analysis (PCA)

### Definition
PCA is a dimensionality reduction technique that transforms many correlated features into fewer principal components.

It attempts to retain as much variance as possible with fewer dimensions.

A dataset may contain 50 related features.

PCA can reduce them to a smaller number of components while retaining important information.

```python
from sklearn.datasets import load_iris

iris = load_iris()

X = iris.data

scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

pca = PCA(n_components=2)

X_pca = pca.fit_transform(X_scaled)

print("Original shape:", X.shape)
print("PCA shape:", X_pca.shape)
print("Explained variance:", pca.explained_variance_ratio_)
```

### Interpretation

The original dataset has four features, while PCA reduces them to two components.

`explained_variance_ratio_` shows how much information each component captures from the original data.

## 13. Recommendation Systems

### Definition
A recommendation system suggests products, movies, songs, or other items that a user may be interested in.

Common approaches include collaborative filtering and content-based filtering.

An e-commerce website can recommend products based on a customer's previous purchases and interactions.

```python
from sklearn.metrics.pairwise import cosine_similarity

user_item = pd.DataFrame({
    "Product_A": [5, 4, 0],
    "Product_B": [4, 5, 1],
    "Product_C": [0, 1, 5]
}, index=["User_1", "User_2", "User_3"])

similarity = cosine_similarity(user_item)

similarity_df = pd.DataFrame(
    similarity,
    index=user_item.index,
    columns=user_item.index
)

print(similarity_df)
```

### Interpretation

Cosine similarity measures how similar users are based on their product interactions.

Users with higher similarity can be used to generate personalized recommendations.
