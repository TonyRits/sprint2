# Correlation

Correlation measures the relationship between two variables.

In this notebook, we will learn correlation, covariance, different types of correlation, Pearson correlation, Spearman correlation, correlation matrix, and heatmap.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from scipy.stats import pearsonr, spearmanr
```

## 1. Correlation

### Step 1 – Understand the Concept

Correlation measures the strength and direction of the relationship between two variables.

The correlation value usually ranges from -1 to +1.

### Step 2 – Demonstrate the Concept

For example, study hours and exam marks may have a positive relationship.

When study hours increase, exam marks may also increase.

```python
study_hours = [1, 2, 3, 4, 5]
marks = [40, 50, 60, 70, 80]

correlation = np.corrcoef(study_hours, marks)[0, 1]

print("Correlation:", correlation)
```

### Interpretation

The correlation is close to +1, showing a strong positive relationship.

Correlation shows association, but it does not prove causation.

## 2. Covariance

### Step 1 – Understand the Concept

Covariance measures how two variables change together.

A positive value means they tend to move in the same direction, while a negative value means they tend to move in opposite directions.

### Step 2 – Demonstrate the Concept

Suppose advertising spending and sales increase together.

Their covariance will generally be positive.

```python
advertising = [10, 20, 30, 40, 50]
sales = [100, 120, 140, 160, 180]

covariance = np.cov(advertising, sales)[0, 1]

print("Covariance:", covariance)
```

### Interpretation

The positive covariance indicates that advertising spending and sales move in the same direction.

Unlike correlation, covariance does not have a fixed range.

## 3. Positive Correlation

### Step 1 – Understand the Concept

Positive correlation means two variables tend to increase or decrease together.

Its correlation value is greater than 0.

### Step 2 – Demonstrate the Concept

For example, as temperature increases, ice cream sales may also increase.

```python
temperature = [20, 22, 24, 26, 28]
ice_cream_sales = [100, 120, 140, 160, 180]

correlation = np.corrcoef(
    temperature,
    ice_cream_sales
)[0, 1]

print("Positive Correlation:", correlation)
```

### Interpretation

The correlation is close to +1, indicating a strong positive relationship.

```python
plt.scatter(temperature, ice_cream_sales)
plt.plot(temperature, ice_cream_sales)
plt.xlabel("Temperature")
plt.ylabel("Ice Cream Sales")
plt.title("Positive Correlation")

plt.show()
```

## 4. Negative Correlation

### Step 1 – Understand the Concept

Negative correlation means one variable tends to increase when the other decreases.

Its correlation value is less than 0.

### Step 2 – Demonstrate the Concept

For example, as the price of a product increases, the number of units sold may decrease.

```python
price = [10, 20, 30, 40, 50]
sales = [100, 80, 60, 40, 20]

correlation = np.corrcoef(price, sales)[0, 1]

print("Negative Correlation:", correlation)
```

### Interpretation

The correlation is close to -1, indicating a strong negative relationship.

```python
plt.scatter(price, sales)
plt.plot(price, sales)
plt.xlabel("Price")
plt.ylabel("Sales")
plt.title("Negative Correlation")

plt.show()
```

## 5. No Correlation

### Step 1 – Understand the Concept

No correlation means there is little or no linear relationship between two variables.

The correlation value is close to 0.

### Step 2 – Demonstrate the Concept

For example, shoe size and the number of books a person owns may have no meaningful linear relationship.

```python
x = [1, 2, 3, 4, 5]
y = [10, 8, 12, 9, 11]

correlation = np.corrcoef(x, y)[0, 1]

print("Correlation:", correlation)
```

### Interpretation

A correlation close to zero indicates little or no linear relationship.

```python
plt.scatter(x, y)
plt.plot(x,y)
plt.xlabel("Variable X")
plt.ylabel("Variable Y")
plt.title("No Correlation")

plt.show()
```

## 6. Pearson Correlation

### Step 1 – Understand the Concept

Pearson correlation measures the strength and direction of a linear relationship between two numerical variables.

Its value ranges from -1 to +1.

### Step 2 – Demonstrate the Concept

If study hours and marks have a strong linear relationship, Pearson correlation will be close to +1.

```python
study_hours = [1, 2, 3, 4, 5]
marks = [40, 50, 60, 70, 80]

pearson_value, p_value = pearsonr(
    study_hours,
    marks
)

print("Pearson Correlation:", pearson_value)
print("P-value:", p_value)
```

### Interpretation

A Pearson correlation close to +1 indicates a strong positive linear relationship.

Pearson correlation is commonly used for numerical continuous variables.

## 7. Spearman Correlation

### Step 1 – Understand the Concept

Spearman correlation measures the strength and direction of a monotonic relationship between two variables.

It is based on the ranks of the observations.

### Step 2 – Demonstrate the Concept

Suppose customers rank products based on quality and satisfaction.

Spearman correlation can measure whether higher quality rankings are generally associated with higher satisfaction rankings.

```python
quality_rank = [1, 2, 3, 4, 5]
satisfaction_rank = [1, 3, 2, 5, 4]

spearman_value, p_value = spearmanr( quality_rank, satisfaction_rank )

print("Spearman Correlation:", spearman_value)
print("P-value:", p_value)
```

### Interpretation

Spearman correlation is useful when the relationship is based on ranks or is not necessarily linear.

It is also less sensitive to extreme values than Pearson correlation in many situations.

## 8. Correlation Matrix

### Step 1 – Understand the Concept

A correlation matrix shows the correlation between multiple numerical variables.

Each cell contains the correlation value between two variables.

### Step 2 – Demonstrate the Concept

Suppose we have data about study hours, marks, attendance, and assignment scores.

A correlation matrix can show the relationship between all these variables.

```python
data = pd.DataFrame({
    "Study_Hours": [1, 2, 3, 4, 5],
    "Marks": [40, 50, 60, 70, 80],
    "Attendance": [60, 65, 75, 85, 90],
    "Assignments": [45, 55, 65, 75, 85]
})

correlation_matrix = data.corr()

print(correlation_matrix)
```

### Interpretation

The matrix shows the correlation value between every pair of numerical variables.

Values close to +1 indicate positive relationships, while values close to -1 indicate negative relationships.

## 9. Heatmap

### Step 1 – Understand the Concept

A heatmap is a visual representation of numerical values using different shades.

A correlation heatmap makes relationships between multiple variables easier to identify.

### Step 2 – Demonstrate the Concept

A correlation matrix can contain many numbers.

A heatmap makes strong positive, negative, and weak relationships easier to see visually.

```python
plt.figure(figsize=(8, 6))

plt.imshow(
    correlation_matrix,
    interpolation="nearest"
)

plt.colorbar()

plt.xticks(
    range(len(correlation_matrix.columns)),
    correlation_matrix.columns,
    rotation=45
)

plt.yticks(
    range(len(correlation_matrix.columns)),
    correlation_matrix.columns
)

plt.title("Correlation Heatmap")
plt.tight_layout()

plt.show()
```

### Interpretation

The heatmap provides a visual view of the correlation matrix.

It helps quickly identify strong positive, strong negative, and weak relationships.
