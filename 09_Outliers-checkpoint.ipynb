# Outliers

Outliers are unusual data points that are very different from most other observations.

In this notebook, we will learn types of outliers, detection methods, IQR, Z-Score, box plots, and scatter plots.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from scipy.stats import zscore
```

## 1. What are Outliers?

### Step 1 – Understand the Concept

An outlier is an observation that is significantly different from the other observations in a dataset.

Outliers can occur because of errors, unusual events, or genuine extreme values.

### Step 2 – Demonstrate the Concept

Consider the values:

10, 12, 11, 13, 12, 50

Most values are around 10–13, while 50 is much larger and may be an outlier.

```python
data = [10, 12, 11, 13, 12, 50]

print("Data:", data)
print("Mean:", np.mean(data))
```

### Interpretation

The value 50 is much higher than the other observations.

Such extreme values can affect statistical calculations like the mean and standard deviation.

## 2. Types of Outliers

### Step 1 – Understand the Concept

Common types of outliers include:

- Univariate Outlier – An unusual value in a single variable.
- Multivariate Outlier – An unusual combination of values across multiple variables.
- Global Outlier – A value that is very different from the entire dataset.
- Contextual Outlier – A value that is unusual only in a specific context or condition.

### Step 2 – Demonstrate the Concept

A single unusually high salary can be a univariate outlier.

A combination of unusual values across multiple features can be a multivariate outlier.

### Interpretation

The type of outlier depends on how the unusual observation differs from the rest of the data.

Understanding the type helps us choose an appropriate detection method.

## 3. Detecting Outliers

### Step 1 – Understand the Concept

Outliers can be detected using statistical methods and visualizations.

Common methods include the IQR method, Z-Score method, box plot, and scatter plot.

### Step 2 – Demonstrate the Concept

Suppose we have customer purchase amounts.

A purchase amount that is extremely different from most purchases may be an outlier.

```python
data = np.array([10, 12, 11, 13, 12, 14, 15, 50])

print("Data:", data)
```

### Interpretation

The value 50 looks unusually high compared with the other observations.

We can use statistical methods to verify whether it is an outlier.

## 4. IQR Method

### Step 1 – Understand the Concept

IQR stands for Interquartile Range.

IQR is calculated as:

IQR = Q3 - Q1

Values below Q1 - 1.5 × IQR or above Q3 + 1.5 × IQR are commonly considered outliers.

### Step 2 – Demonstrate the Concept

The IQR method focuses on the middle 50% of the data.

It is useful because it is less affected by extreme values.

```python
data = np.array([10, 12, 11, 13, 12, 14, 15, 50])

Q1 = np.percentile(data, 25)
Q3 = np.percentile(data, 75)

IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

outliers = data[
    (data < lower_bound) |
    (data > upper_bound)
]

print("Q1:", Q1)
print("Q3:", Q3)
print("IQR:", IQR)
print("Lower Bound:", lower_bound)
print("Upper Bound:", upper_bound)
print("Outliers:", outliers)
```

### Interpretation

The IQR method identifies values outside the lower and upper bounds.

In this example, 50 is identified as an outlier.

## 5. Z-Score Method

### Step 1 – Understand the Concept

Z-Score measures how far a value is from the mean in terms of standard deviations.

A common rule is that values with an absolute Z-Score greater than 3 may be treated as outliers.

### Step 2 – Demonstrate the Concept

If a value has a Z-Score of 0, it is equal to the mean.

A large positive or negative Z-Score means the value is far from the mean.

```python
data = np.array([10, 12, 11, 13, 12, 14, 15, 50])

z_scores = zscore(data)

outliers = data[np.abs(z_scores) > 3]

print("Z-Scores:", z_scores)
print("Outliers:", outliers)
```

### Interpretation

The Z-Score method measures how far each observation is from the mean.

Values with an absolute Z-Score greater than 3 are commonly considered potential outliers.

## 6. Box Plot

### Step 1 – Understand the Concept

A box plot shows the distribution of numerical data using the minimum, Q1, median, Q3, and maximum.

It can also visually identify potential outliers.

### Step 2 – Demonstrate the Concept

A box plot represents the middle 50% of the data using a box.

Values outside the whiskers are usually displayed as potential outliers.

```python
data = [10, 12, 11, 13, 12, 14, 15, 50]

plt.boxplot(data)

plt.ylabel("Value")
plt.title("Box Plot for Outlier Detection")

plt.show()
```

### Interpretation

The box plot makes the extreme value visually visible.

Box plots are useful for quickly identifying potential outliers in numerical data.

## 7. Scatter Plot

### Step 1 – Understand the Concept

A scatter plot displays the relationship between two numerical variables.

It can help identify unusual observations that are far away from the main pattern.

### Step 2 – Demonstrate the Concept

Suppose we analyze advertising spending and sales.

If one observation is very far from the general pattern, it may be a potential outlier.

```python
advertising = [10, 20, 30, 40, 50, 60]
sales = [100, 120, 140, 160, 180, 400]

plt.scatter(advertising, sales)

plt.xlabel("Advertising Spending")
plt.ylabel("Sales")
plt.title("Scatter Plot for Outlier Detection")

plt.show()
```

### Interpretation

Most observations follow a general pattern, while the point with sales of 400 is far from the others.

This may indicate a potential outlier.
