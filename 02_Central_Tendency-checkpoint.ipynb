# Measures of Central Tendency

## Sprint 2 – Statistics & Mathematics for AI/ML Engineers

Measures of central tendency help us find a typical or central value
in a dataset. The main measures are Mean, Median, and Mode.

```python
import numpy as np
import pandas as pd
```

## 1. Mean

### Step 1 – Understand the Concept

Mean is the average value of a dataset. It is calculated by adding
all values and dividing the total by the number of values.

### Formula

Mean = Sum of all values / Number of values

### Step 2 – Demonstrate the Concept

Suppose the daily sales are 100, 120, 150, 130, and 200.

Mean = (100 + 120 + 150 + 130 + 200) / 5 = 140.

The average daily sales are 140.

### Business Example

A company can use mean sales to understand its average daily revenue.

### AI/ML Example

Mean can be used to summarize numerical features and to replace
missing numerical values during data preprocessing.

```python
sales = [100, 120, 150, 130, 200]

mean_value = np.mean(sales)

print("Sales:", sales)
print("Mean:", mean_value)
```

### Interpretation

The mean sales value is 140. This represents the average sales
across the five days.

### Advantages
- Easy to calculate and understand.
- Uses all values in the dataset.

### Limitations
- Sensitive to extreme values or outliers.

## 2. Median

### Step 1 – Understand the Concept

Median is the middle value when data is arranged in ascending or descending order.
It is useful when the dataset contains extreme values.

### Step 2 – Demonstrate the Concept

Consider the salaries:

30,000, 35,000, 40,000, 45,000, 200,000

The median is 40,000 because it is the middle value.

### Business Example

A company can use median salary to understand the typical employee salary
without allowing a very high salary to strongly affect the result.

### AI/ML Example

Median can be used for missing-value imputation when numerical data contains outliers.

```python
salaries = [30000, 35000, 40000, 45000, 200000]

median_value = np.median(salaries)

print("Salaries:", salaries)
print("Median:", median_value)
```

### Interpretation

The median salary is 40,000. The extreme value of 200,000 does not
strongly affect the median.

### Advantages
- Less affected by outliers.
- Useful for skewed data.

### Limitations
- Does not directly use the magnitude of every value.

## 3. Mode

### Step 1 – Understand the Concept

Mode is the value that occurs most frequently in a dataset.
A dataset can have one mode, multiple modes, or no mode.

### Step 2 – Demonstrate the Concept

Suppose the most frequently purchased product sizes are:

S, M, M, L, M, S

The mode is M because it occurs most often.

### Business Example

A clothing company can identify the most commonly purchased size.

### AI/ML Example

Mode can be useful for filling missing values in categorical features.

```python
sizes = ["S", "M", "M", "L", "M", "S"]

mode_value = pd.Series(sizes).mode()[0]

print("Sizes:", sizes)
print("Mode:", mode_value)
```

### Interpretation

The mode is M because it appears more frequently than the other sizes.

### Advantages
- Easy to understand.
- Works with categorical data.

### Limitations
- There may be multiple modes.
- It may not exist when all values occur equally often.

## 4. Weighted Mean

### Step 1 – Understand the Concept

Weighted mean is an average where different values have different levels of importance.
Values with higher weights have a greater effect on the final result.

### Formula

Weighted Mean = Σ(value × weight) / Σ(weight)

### Step 2 – Demonstrate the Concept

Suppose a student's final grade is based on:

Assignment = 80 with weight 20%
Exam = 90 with weight 80%

Weighted Mean = (80 × 0.20) + (90 × 0.80)
              = 88

### Business Example

A company may calculate an employee performance score by giving
different weights to productivity, quality, and attendance.

### AI/ML Example

Weighted averages can be used when some observations or measurements
are considered more important than others.

```python
scores = [80, 90]
weights = [0.20, 0.80]

weighted_mean = np.average(scores, weights=weights)

print("Weighted Mean:", weighted_mean)
```

### Interpretation

The weighted mean is 88 because the exam has a higher weight than the assignment.

### Advantages
- Gives more importance to important values.
- Useful when observations have different importance.

### Limitations
- Choosing inappropriate weights can produce misleading results.

## 5. Geometric Mean

### Step 1 – Understand the Concept

Geometric mean is an average calculated by multiplying values
and taking the nth root of the product.

### Formula

Geometric Mean = (x₁ × x₂ × ... × xₙ)^(1/n)

### Step 2 – Demonstrate the Concept

Consider the values 2, 8, and 32.

Geometric Mean = (2 × 8 × 32)^(1/3)
               = 8

### Business Example

Geometric mean can be used to calculate average growth rates
over multiple periods.

### AI/ML Example

It can be useful when working with multiplicative changes,
growth rates, or ratios.

```python
values = [2, 8, 32]

geometric_mean = np.prod(values) ** (1 / len(values))

print("Values:", values)
print("Geometric Mean:", geometric_mean)
```

### Interpretation

The geometric mean is 8. It is especially useful for values
that represent growth or multiplicative changes.

### Advantages
- Useful for growth rates and ratios.
- Less influenced by very large values than arithmetic mean in some contexts.

### Limitations
- Cannot normally be used directly with negative values.
- More difficult to interpret than arithmetic mean.

## 6. Harmonic Mean

### Step 1 – Understand the Concept

Harmonic mean is an average that is useful for rates and ratios.
It is calculated as the number of values divided by the sum of their reciprocals.

### Formula

Harmonic Mean = n / Σ(1/x)

### Step 2 – Demonstrate the Concept

Suppose a vehicle travels:

60 km/h for one part of a journey
40 km/h for another part of the journey

For equal distances, the harmonic mean gives the appropriate average speed.

Harmonic Mean = 2 / (1/60 + 1/40) = 48 km/h

### Business Example

It can be used when averaging rates such as speed, efficiency,
or price-to-performance ratios.

### AI/ML Example

Harmonic mean is used in evaluation metrics such as the F1-score,
which combines Precision and Recall.

```python
speeds = [60, 40]

harmonic_mean = len(speeds) / sum(1 / x for x in speeds)

print("Speeds:", speeds)
print("Harmonic Mean:", harmonic_mean)
```

### Interpretation

The harmonic mean is 48 km/h for the two equal-distance speeds.
It gives more importance to smaller values.

### Advantages
- Useful for rates and ratios.
- Used in important ML metrics such as F1-score.

### Limitations
- Cannot be calculated when a value is zero.
- Not suitable for every type of dataset.
