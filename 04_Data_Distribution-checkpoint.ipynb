# Data Distribution

## Sprint 2 – Statistics & Mathematics for AI/ML Engineers

Data distribution describes how values are spread across a dataset.
Understanding distributions helps us identify patterns, skewness,
variability, and unusual observations.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

## 1. Normal Distribution

### Step 1 – Understand the Concept

A normal distribution is a continuous distribution where most values
are concentrated around the mean and the distribution is symmetric.

It has a bell-shaped curve, with the mean, median, and mode at the center.

### Formula

f(x) = (1 / (σ√(2π))) × e^(-(x-μ)² / (2σ²))

### Step 2 – Demonstrate the Concept

For example, the heights of a large group of people may approximately
follow a normal distribution, with most people having heights near the average.

### Business Example

A company can analyze employee performance scores to understand
whether most scores are concentrated around an average value.

### AI/ML Example

Normal distributions are important in statistical modeling and are
used in algorithms and techniques such as Gaussian Naive Bayes.

```python
data = np.random.normal(loc=50, scale=10, size=1000)

plt.hist(data, bins=30, density=True)
plt.xlabel("Value")
plt.ylabel("Density")
plt.title("Normal Distribution")
plt.show()
```

### Interpretation

The histogram forms an approximately bell-shaped curve.
Most observations are near the mean of 50, while fewer observations
appear farther away from the mean.

## 2. Uniform Distribution

### Step 1 – Understand the Concept

In a uniform distribution, values within a specified range have
approximately equal probability of occurring.

For a continuous uniform distribution, the probability density is constant.

### Formula

f(x) = 1 / (b - a), for a ≤ x ≤ b

### Step 2 – Demonstrate the Concept

Suppose a random number is selected between 1 and 10.
Every value within this range has an equal chance of being selected.

### Business Example

A simulation may randomly assign customers to groups with equal probability.

### AI/ML Example

Uniform random values can be used when initializing parameters
or generating simulated data.

```python
data = np.random.uniform(1, 10, 1000)

plt.hist(data, bins=20)
plt.xlabel("Value")
plt.ylabel("Frequency")
plt.title("Uniform Distribution")
plt.show()
```

### Interpretation

The values are spread relatively evenly across the specified range.
There is no strong concentration around a particular value.

### Interpretation

The values are spread relatively evenly across the specified range.
There is no strong concentration around a particular value.

## 3. Gaussian Distribution

### Step 1 – Understand the Concept

Gaussian distribution is another name commonly used for the normal distribution.
It is a symmetric, bell-shaped distribution defined by its mean and standard deviation.

### Step 2 – Demonstrate the Concept

If exam scores are approximately Gaussian, most students will score
near the average, while very low and very high scores occur less frequently.

### Business Example

A company can analyze measurements such as product dimensions
when the measurements approximately follow a Gaussian distribution.

### AI/ML Example

Gaussian assumptions are used in models such as Gaussian Naive Bayes.

```python
mean = 100
std = 15

data = np.random.normal(mean, std, 1000)

plt.hist(data, bins=30)
plt.xlabel("Value")
plt.ylabel("Frequency")
plt.title("Gaussian Distribution")
plt.show()
```

### Interpretation

The generated data is centered around 100 and has a standard deviation
of approximately 15.

Gaussian and normal distribution refer to the same distribution in this context.

## 4. Left Skew

### Step 1 – Understand the Concept

A left-skewed distribution has a longer tail on the left side.
Most observations are concentrated toward the higher values.

In a typical left-skewed distribution:

Mean < Median < Mode

### Step 2 – Demonstrate the Concept

Suppose most students score between 70 and 100, but a few students
score very low. The low scores create a longer tail toward the left.

### Business Example

If most customers spend a high amount but a few spend very little,
the spending distribution may be left-skewed.

### AI/ML Example

Understanding skewness helps decide whether a feature needs
transformation before model training.

```python
data = np.random.beta(5, 2, 1000) * 100

plt.hist(data, bins=30)
plt.xlabel("Value")
plt.ylabel("Frequency")
plt.title("Left-Skewed Distribution")
plt.show()
```

### Interpretation

Most values are concentrated toward the higher end,
while the tail extends toward the lower values.

## 5. Right Skew

### Step 1 – Understand the Concept

A right-skewed distribution has a longer tail on the right side.
Most observations are concentrated toward the lower values.

In a typical right-skewed distribution:

Mode < Median < Mean

### Step 2 – Demonstrate the Concept

Income is often used as an example because many people may have
moderate incomes while a smaller number of people have very high incomes.

### Business Example

Customer spending may be right-skewed when most customers spend
moderate amounts but a few customers spend extremely large amounts.

### AI/ML Example

Right-skewed features may be transformed using methods such as
log transformation to make the distribution more suitable for some models.

```python
data = np.random.exponential(scale=2, size=1000)

plt.hist(data, bins=30)
plt.xlabel("Value")
plt.ylabel("Frequency")
plt.title("Right-Skewed Distribution")
plt.show()
```

### Interpretation

Most values are concentrated near the lower end,
while a long tail extends toward higher values.

## 6. Symmetric Distribution

### Step 1 – Understand the Concept

A distribution is symmetric when the left and right sides have
similar shapes around the center.

For a perfectly normal distribution, mean = median = mode.

### Step 2 – Demonstrate the Concept

A bell-shaped normal distribution is an example of a symmetric distribution.
The values are distributed similarly on both sides of the center.

### Business Example

A measurement process with errors equally likely in both directions
may produce approximately symmetric data.

### AI/ML Example

Checking symmetry helps understand whether a feature follows
a balanced distribution or has skewness.

```python
data = np.random.normal(50, 10, 1000)

plt.hist(data, bins=30)
plt.xlabel("Value")
plt.ylabel("Frequency")
plt.title("Symmetric Distribution")
plt.show()
```

### Interpretation

The distribution is approximately symmetric around its mean.
Both sides have a similar shape.

## 7. Kurtosis

### Step 1 – Understand the Concept

Kurtosis describes the shape of the tails of a distribution
and helps identify how much extreme values may occur.

A distribution can have lower, similar, or higher tail heaviness
compared with a normal distribution.

### Step 2 – Demonstrate the Concept

Two datasets can have similar means and standard deviations
but different numbers of extreme observations.

Kurtosis helps identify differences in their tail behavior.

### Business Example

A financial company can analyze kurtosis to understand
the possibility of extreme changes in financial returns.

### AI/ML Example

Kurtosis can help identify features that contain unusually
heavy tails or extreme observations.

```python
from scipy.stats import kurtosis

data = np.random.normal(50, 10, 1000)

kurtosis_value = kurtosis(data)

print("Kurtosis:", kurtosis_value)
```

### Interpretation

The calculated kurtosis describes the tail behavior of the generated data.
A value close to zero here is expected because SciPy's default calculation
uses excess kurtosis, where a normal distribution has approximately zero.

## 8. Empirical Rule (68–95–99.7)

### Step 1 – Understand the Concept

For an approximately normal distribution:

- About 68% of values fall within 1 standard deviation of the mean.
- About 95% fall within 2 standard deviations.
- About 99.7% fall within 3 standard deviations.

This is called the Empirical Rule.

### Step 2 – Demonstrate the Concept

Suppose exam scores have:

Mean = 70
Standard Deviation = 10

Approximately:

68% → 60 to 80
95% → 50 to 90
99.7% → 40 to 100

### Business Example

A company can use this rule to understand the expected range
of normally distributed measurements.

### AI/ML Example

The rule can help identify unusually distant observations
when a feature approximately follows a normal distribution.

```python
mean = 70
std = 10

range_1 = (mean - std, mean + std)
range_2 = (mean - 2 * std, mean + 2 * std)
range_3 = (mean - 3 * std, mean + 3 * std)

print("Within 1 Standard Deviation:", range_1)
print("Within 2 Standard Deviations:", range_2)
print("Within 3 Standard Deviations:", range_3)
```

### Interpretation

For a normal distribution with mean 70 and standard deviation 10,
most observations are expected to fall within 60–80.

Almost all observations are expected to fall within 40–100.
