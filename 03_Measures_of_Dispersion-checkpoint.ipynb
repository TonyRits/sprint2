# Measures of Dispersion

## Sprint 2 – Statistics & Mathematics for AI/ML Engineers

Measures of dispersion describe how spread out the values in a dataset are.
They help us understand the variability and consistency of data.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

## 1. Range

### Step 1 – Understand the Concept

Range is the difference between the maximum and minimum values in a dataset.
It gives a simple idea of how widely the data is spread.

### Formula

Range = Maximum Value - Minimum Value

### Step 2 – Demonstrate the Concept

Suppose the daily sales are:

100, 120, 150, 180, 200

Range = 200 - 100 = 100

### Business Example

A company can use the range of daily sales to understand the difference
between its lowest and highest sales.

### AI/ML Example

Range can help understand the spread of a numerical feature during EDA.

```python
sales = [100, 120, 150, 180, 200]

data_range = np.max(sales) - np.min(sales)

print("Sales:", sales)
print("Range:", data_range)
```

### Interpretation

The range is 100, meaning the difference between the highest and lowest
sales value is 100.

### Advantages
- Very easy to calculate.
- Easy to understand.

### Limitations
- Uses only the minimum and maximum values.
- Highly affected by outliers.

## 2. Variance

### Step 1 – Understand the Concept

Variance measures how far the values in a dataset are spread from the mean.
A higher variance means the data has more variability.

### Formula

Population Variance = Σ(x - μ)² / N

Sample Variance = Σ(x - x̄)² / (n - 1)

### Step 2 – Demonstrate the Concept

Consider the values:

2, 4, 6

Mean = 4

The values are spread around the mean, so variance measures
how much they differ from 4.

### Business Example

A company can use variance to understand how much monthly sales
change from their average value.

### AI/ML Example

Variance is important in feature analysis and is also used in
algorithms such as PCA.

```python
data = [2, 4, 6]

population_variance = np.var(data)
sample_variance = np.var(data, ddof=1)

print("Population Variance:", population_variance)
print("Sample Variance:", sample_variance)
```

### Interpretation

Population variance describes the spread of the complete population,
while sample variance estimates the variance of a population using a sample.

### Advantages
- Uses every value in the dataset.
- Useful for measuring variability.

### Limitations
- The unit is squared, making interpretation less direct.
- Sensitive to outliers.

## 3. Standard Deviation

### Step 1 – Understand the Concept

Standard deviation measures the typical spread of values around the mean.
It is the square root of variance and is expressed in the same unit as the data.

### Formula

Standard Deviation = √Variance

### Step 2 – Demonstrate the Concept

Suppose the marks are:

70, 75, 80, 85, 90

A small standard deviation means the marks are relatively close together.

### Business Example

A company can use standard deviation to measure variation in
monthly sales or customer spending.

### AI/ML Example

Standard deviation is used in feature scaling, normalization,
and detecting unusual observations.

```python
marks = [70, 75, 80, 85, 90]

std = np.std(marks)

print("Marks:", marks)
print("Standard Deviation:", std)
```

### Interpretation

The standard deviation tells us how much the marks typically vary
around the average mark.

### Advantages
- Easy to interpret because it uses the original unit.
- Uses all observations.

### Limitations
- Sensitive to outliers.
- Can be misleading for highly skewed data.

## 4. Quartiles

### Step 1 – Understand the Concept

Quartiles divide an ordered dataset into four parts.
Q1 represents the 25th percentile, Q2 represents the median,
and Q3 represents the 75th percentile.

### Step 2 – Demonstrate the Concept

Consider:

10, 20, 30, 40, 50, 60, 70

Q1 represents the point below which approximately 25% of the data falls.
Q2 is the median, and Q3 represents the point below which approximately 75% falls.

### Business Example

Quartiles can be used to divide customers into low, medium,
and high spending groups.

### AI/ML Example

Quartiles are used in EDA and are important for detecting outliers.

```python
data = [10, 20, 30, 40, 50, 60, 70]

q1 = np.percentile(data, 25)
q2 = np.percentile(data, 50)
q3 = np.percentile(data, 75)

print("Q1:", q1)
print("Q2:", q2)
print("Q3:", q3)
```

### Interpretation

Q1, Q2, and Q3 divide the dataset into four approximate sections.
They help us understand where values are located within the distribution.

## 5. Percentiles

### Step 1 – Understand the Concept

A percentile indicates the value below which a certain percentage
of observations fall.

For example, the 75th percentile means approximately 75% of values
are at or below that value.

### Step 2 – Demonstrate the Concept

Suppose the 90th percentile of test scores is 95.
This means approximately 90% of the scores are at or below 95.

### Business Example

Companies can use percentiles to identify high-spending customers.

### AI/ML Example

Percentiles can help understand feature distributions and detect extreme values.

```python
scores = [50, 60, 65, 70, 75, 80, 85, 90, 95, 100]

p90 = np.percentile(scores, 90)

print("90th Percentile:", p90)
```

### Interpretation

The 90th percentile identifies a value near the upper end of the score distribution.
Percentiles are useful for comparing observations within a dataset.

## 6. Interquartile Range (IQR)

### Step 1 – Understand the Concept

IQR measures the spread of the middle 50% of a dataset.
It is calculated as the difference between Q3 and Q1.

### Formula

IQR = Q3 - Q1

### Step 2 – Demonstrate the Concept

Suppose:

Q1 = 20
Q3 = 60

IQR = 60 - 20 = 40

This means the middle 50% of the data covers a range of 40 units.

### Business Example

IQR can be used to understand the spread of customer spending
while reducing the effect of extreme customers.

### AI/ML Example

IQR is commonly used to detect outliers during data preprocessing.

```python
data = [10, 20, 30, 40, 50, 60, 70, 100]

q1 = np.percentile(data, 25)
q3 = np.percentile(data, 75)

iqr = q3 - q1

print("Q1:", q1)
print("Q3:", q3)
print("IQR:", iqr)
```

### Interpretation

IQR represents the spread of the middle 50% of the observations.
It is less affected by extreme values than range and standard deviation.

### Advantages
- Resistant to outliers.
- Useful for skewed data.
- Commonly used for outlier detection.

### Limitations
- Does not use every observation.
- Does not describe the exact shape of the distribution.

## 7. Coefficient of Variation

### Step 1 – Understand the Concept

Coefficient of Variation (CV) measures relative variability compared with the mean.
It is useful for comparing the variation of datasets with different scales.

### Formula

CV = (Standard Deviation / Mean) × 100

### Step 2 – Demonstrate the Concept

Dataset A:

Mean = 100
Standard Deviation = 10

CV = (10 / 100) × 100 = 10%

Dataset B:

Mean = 200
Standard Deviation = 40

CV = (40 / 200) × 100 = 20%

Dataset B has greater relative variability.

### Business Example

A company can compare the consistency of sales for two different products.

### AI/ML Example

CV can help compare variability between features that have different scales.

```python
mean = 100
std = 10

cv = (std / mean) * 100

print("Coefficient of Variation:", cv, "%")
```

### Interpretation

The coefficient of variation is 10%.
This means the standard deviation is 10% of the mean.

A higher CV generally indicates greater relative variability.
