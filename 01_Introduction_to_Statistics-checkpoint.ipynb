# Introduction to Statistics

## Sprint 2 – Statistics & Mathematics for AI/ML Engineers

This notebook introduces the basic concepts of statistics and explains
how statistics is used to understand, analyze, and interpret data in AI/ML.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

## 1. What is Statistics?

### Step 1 – Understand the Concept

Statistics is the process of collecting, organizing, analyzing, and interpreting data.
It helps us understand patterns and make decisions based on data.

### Step 2 – Demonstrate the Concept

For example, a company records the daily sales of a product.
By calculating the average sales, the company can understand its typical daily performance.

This demonstrates how statistics helps convert raw data into useful information.

```python
sales = [100, 120, 150, 130, 170]

average_sales = np.mean(sales)

print("Daily Sales:", sales)
print("Average Sales:", average_sales)
```

### Interpretation

The average sales represent the typical daily sales of the product.
In AI/ML, statistics is used to understand data before building models.

## 2. Why Statistics is Important in AI/ML

### Step 1 – Understand the Concept

Statistics helps AI/ML engineers understand data, identify patterns,
detect outliers, handle missing values, and evaluate machine learning models.

### Step 2 – Demonstrate the Concept

Suppose the ages of customers are [20, 22, 25, 24, 80].
The value 80 may be unusual compared with the other values.

Statistics helps identify such unusual values before training a model.

```python
ages = [20, 22, 25, 24, 80]

print("Mean Age:", np.mean(ages))
print("Median Age:", np.median(ages))
```

### Interpretation

The mean is affected by the unusually high value, while the median
is more resistant to the outlier. This shows why statistical measures
are useful during data preprocessing.

## 3. Statistics vs Machine Learning

### Step 1 – Understand the Concept

Statistics mainly focuses on understanding data, relationships, and uncertainty.
Machine Learning uses data to learn patterns and make predictions or decisions.

### Step 2 – Demonstrate the Concept

Statistics can be used to find the average house price in an area.
Machine Learning can use house features such as area and bedrooms to predict house prices.

```python
house_prices = [250000, 280000, 300000, 270000, 350000]

average_price = np.mean(house_prices)

print("Average House Price:", average_price)
```

### Interpretation

The calculation summarizes the existing data using statistics.
A machine learning model would go further by learning from features
and predicting prices for new houses.

## 4. Descriptive Statistics

### Step 1 – Understand the Concept

Descriptive statistics summarizes the main characteristics of a dataset.
Mean, median, mode, range, and standard deviation are common descriptive measures.

### Step 2 – Demonstrate the Concept

For student marks [60, 70, 70, 80, 90], we can calculate the mean,
median, and standard deviation to understand the overall performance.

```python
marks = [60, 70, 70, 80, 90]

print("Mean:", np.mean(marks))
print("Median:", np.median(marks))
print("Standard Deviation:", np.std(marks))
```

### Interpretation

These values provide a summary of the students' marks.
Descriptive statistics is commonly used during Exploratory Data Analysis (EDA).m

## 5. Inferential Statistics

### Step 1 – Understand the Concept

Inferential statistics uses information from a sample to make conclusions
about a larger population.

### Step 2 – Demonstrate the Concept

A company may survey 100 customers instead of asking every customer.
The results from those 100 customers can be used to understand the larger customer population.

```python
sample = [72, 75, 78, 80, 76, 74, 79]

sample_mean = np.mean(sample)

print("Sample Mean:", sample_mean)
```

### Interpretation

The sample mean gives an estimate of the typical value in the population.
Inferential statistics is useful when collecting data from the entire population is difficult.

## 6. Population

### Step 1 – Understand the Concept

A population is the complete group of individuals or objects that we want to study.
It represents the entire group relevant to a statistical problem.

### Step 2 – Demonstrate the Concept

If we want to study the marks of every student in a college,
all students in that college form the population.

```python
population_marks = [65, 70, 75, 80, 85, 90]

print("Population:", population_marks)
print("Population Size:", len(population_marks))
```

### Interpretation

The list contains all observations being considered in the study.
In AI/ML, the population represents the complete data group we are interested in.

## 7. Sample

### Step 1 – Understand the Concept

A sample is a smaller group selected from a population.
It is used when studying the entire population is difficult or expensive.

### Step 2 – Demonstrate the Concept

If a college has 10,000 students, we may select 500 students
to understand their average study hours.

```python
population = np.arange(1, 10001)

sample = np.random.choice(population, size=500, replace=False)

print("Population Size:", len(population))
print("Sample Size:", len(sample))
```

### Interpretation

The sample contains 500 observations selected from the population of 10,000.
A properly selected sample can provide useful information about the population.

## 8. Census

### Step 1 – Understand the Concept

A census collects information from every member of the population.
It does not use only a subset of the population.

### Step 2 – Demonstrate the Concept

If a company has 1,000 employees and collects information from
all 1,000 employees, this is a census.

```python
employees = np.arange(1, 1001)

print("Total Employees Surveyed:", len(employees))
```

### Interpretation

All members of the population are included in the analysis.
A census can provide complete information but may require more time and resources.

## 9. Sampling

### Step 1 – Understand the Concept

Sampling is the process of selecting a smaller group from a population.
The selected sample is used to study or estimate characteristics of the population.

### Step 2 – Demonstrate the Concept

Instead of checking every product manufactured by a company,
quality engineers may inspect a smaller random group of products.

```python
products = np.arange(1, 101)

sample = np.random.choice(products, size=10, replace=False)

print("Selected Sample:", sample)
```

### Interpretation

Ten products were randomly selected from 100 products.
Sampling reduces the amount of data that needs to be inspected.

## 10. Variables

### Step 1 – Understand the Concept

A variable is a characteristic or value that can change between observations.
Examples include age, salary, height, gender, and temperature.

### Step 2 – Demonstrate the Concept

For students, age and marks can have different values for different students.
Therefore, age and marks are variables.

```python
students = pd.DataFrame({
    "Age": [20, 21, 22, 20],
    "Marks": [75, 82, 68, 90]
})

print(students)
```

### Interpretation

Age and Marks are variables because their values differ between observations.
In ML, variables are commonly used as inputs or targets.

## 11. Features

### Step 1 – Understand the Concept

Features are input variables used by a machine learning model to learn patterns.
They provide information that helps the model make predictions.

### Step 2 – Demonstrate the Concept

To predict a house price, features could include area, number of bedrooms,
location, and house age.

```python
houses = pd.DataFrame({
    "Area": [1000, 1500, 2000],
    "Bedrooms": [2, 3, 4],
    "Age": [10, 5, 2]
})

print(houses)
```

### Interpretation

Area, Bedrooms, and Age are features that can be given to a machine learning model.
The model can learn how these features relate to house prices.

## 12. Observation

### Step 1 – Understand the Concept

An observation is a single record or data point in a dataset.
Each observation represents one individual item, person, transaction, or event.

### Step 2 – Demonstrate the Concept

In a student dataset, each student's information represents one observation.

```python
students = pd.DataFrame({
    "Name": ["Tony", "Alex", "John"],
    "Age": [22, 21, 23],
    "Marks": [85, 78, 90]
})

print(students.iloc[0])
```

### Interpretation

The first row represents one observation.
In machine learning, each observation is usually one training or testing example.

## 13. Dataset

### Step 1 – Understand the Concept

A dataset is a collection of related observations and variables.
It is commonly represented as a table containing rows and columns.

### Step 2 – Demonstrate the Concept

A customer dataset may contain customer age, income, location,
and purchase amount.

```python
customers = pd.DataFrame({
    "Age": [22, 30, 35, 28],
    "Income": [25000, 40000, 55000, 35000],
    "Purchase": [2000, 4500, 6000, 3000]
})

print(customers)
```

### Interpretation

The DataFrame contains multiple observations and variables.
Machine learning models learn patterns from datasets.

## 14. Quantitative Data

### Step 1 – Understand the Concept

Quantitative data represents numerical values that can be measured or counted.
Examples include age, salary, height, weight, and sales.

### Step 2 – Demonstrate the Concept

A company's monthly sales of 10,000, 12,000, and 15,000 are quantitative data
because the values are numerical.

```python
sales = [10000, 12000, 15000]

print("Sales:", sales)
print("Average Sales:", np.mean(sales))
```

### Interpretation

Sales values are numerical and can be analyzed using mathematical and statistical methods.
Quantitative features are commonly used in machine learning models.

## 15. Qualitative Data

### Step 1 – Understand the Concept

Qualitative data describes categories or characteristics rather than numerical measurements.
Examples include gender, product category, city, and customer type.

### Step 2 – Demonstrate the Concept

A customer dataset may contain categories such as Gold, Silver, and Bronze.
These categories represent qualitative information.

```python
customer_type = ["Gold", "Silver", "Gold", "Bronze", "Silver"]

print(customer_type)
print(pd.Series(customer_type).value_counts())
```

### Interpretation

The values represent categories rather than measurements.
In ML, qualitative data often needs to be converted into numerical form using encoding techniques.

## 16. Continuous Data

### Step 1 – Understand the Concept

Continuous data can take any value within a range.
Examples include height, weight, temperature, and time.

### Step 2 – Demonstrate the Concept

A person's height could be 170 cm, 170.5 cm, or 170.55 cm.
This shows that continuous values can contain decimals.

```python
heights = [165.2, 170.5, 172.8, 168.4]

print("Heights:", heights)
print("Average Height:", np.mean(heights))
```

### Interpretation

Height can take many possible values within a range.
Continuous features are common in numerical machine learning datasets.

## 17. Discrete Data

### Step 1 – Understand the Concept

Discrete data consists of countable values, usually whole numbers.
Examples include number of customers, number of products, and number of purchases.

### Step 2 – Demonstrate the Concept

A customer may purchase 1, 2, or 3 products.
A value such as 2.5 products does not make sense in this situation.

```python
products_purchased = [1, 2, 3, 2, 5]

print("Products Purchased:", products_purchased)
print("Total Products:", sum(products_purchased))
```

### Interpretation

The number of products purchased is countable and therefore discrete.
Discrete variables are frequently used as features in ML datasets.

## 18. Structured Data

### Step 1 – Understand the Concept

Structured data is organized in a fixed format, usually rows and columns.
Examples include Excel files, CSV files, and relational databases.

### Step 2 – Demonstrate the Concept

A customer table containing Customer ID, Age, Income, and Purchase
is an example of structured data.

```python
data = pd.DataFrame({
    "Customer_ID": [1, 2, 3],
    "Age": [25, 30, 35],
    "Purchase": [2000, 3500, 5000]
})

print(data)
```

### Interpretation

The data is arranged into clearly defined rows and columns.
Structured data is easy to process using Pandas and SQL.

## 19. Unstructured Data

### Step 1 – Understand the Concept

Unstructured data does not follow a fixed table-like format.
Examples include images, videos, audio files, emails, and text documents.

### Step 2 – Demonstrate the Concept

A customer review such as "The product quality is excellent"
is text data and does not naturally fit into fixed numerical columns.

```python
customer_review = "The product quality is excellent"

print("Customer Review:", customer_review)
print("Number of Characters:", len(customer_review))
```

### Interpretation

The review is unstructured text data.
AI/ML techniques such as Natural Language Processing can convert text into useful features.
