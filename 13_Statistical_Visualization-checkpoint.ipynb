# Statistical Visualization

Statistical visualization helps us understand patterns, distributions, relationships, and trends in data.

Different charts are useful for different types of data.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

## 1. Histogram

### Purpose
A histogram shows the distribution of numerical data using intervals called bins.

### When to use
Use it to understand data distribution, spread, skewness, and frequency.

### Advantages
- Easy to understand
- Shows distribution clearly
- Helps identify skewness and unusual values

### Limitations
- Bin size can affect the appearance
- Not suitable for categorical data

```python
data = [10, 12, 15, 15, 18, 20, 20, 21, 22, 25, 28, 30]

plt.hist(data, bins=5)
plt.xlabel("Values")
plt.ylabel("Frequency")
plt.title("Histogram")
plt.show()
```

### Interpretation

The bars show how frequently values occur within each range.

Higher bars represent ranges containing more observations.

## 2. Bar Chart

### Purpose
A bar chart compares values across different categories.

### When to use
Use it for categorical data such as product sales, departments, or regions.

### Advantages
- Easy category comparison
- Simple to understand
- Works well for business reports

### Limitations
- Not suitable for showing continuous distributions
- Too many categories can make it difficult to read

```python
products = ["Laptop", "Phone", "Tablet", "Watch"]
sales = [120, 200, 150, 100]

plt.bar(products, sales)
plt.xlabel("Product")
plt.ylabel("Sales")
plt.title("Product Sales")
plt.show()
```

### Interpretation

The Phone category has the highest sales, while Watch has the lowest sales.

## 3. Line Chart

### Purpose
A line chart shows changes or trends over an ordered sequence.

### When to use
Use it for time-series data such as monthly sales or yearly revenue.

### Advantages
- Clearly shows trends
- Useful for time-series analysis
- Easy to compare changes

### Limitations
- Not ideal for unrelated categories
- Too many lines can make the chart confusing

```python
months = ["Jan", "Feb", "Mar", "Apr", "May"]
sales = [100, 120, 115, 140, 160]

plt.plot(months, sales, marker="o")
plt.xlabel("Month")
plt.ylabel("Sales")
plt.title("Monthly Sales Trend")
plt.show()
```

### Interpretation

Sales generally increase from January to May, with a small decrease in March.

## 4. Scatter Plot

### Purpose
A scatter plot shows the relationship between two numerical variables.

### When to use
Use it to identify correlation, patterns, clusters, and possible outliers.

### Advantages
- Shows relationships clearly
- Helps identify correlation
- Can reveal outliers

### Limitations
- Difficult to interpret with very large datasets
- Correlation does not prove causation

```python
advertising = [10, 20, 30, 40, 50]
sales = [15, 25, 32, 42, 55]

plt.scatter(advertising, sales)
plt.xlabel("Advertising Spend")
plt.ylabel("Sales")
plt.title("Advertising vs Sales")
plt.show()
```

### Interpretation

The points show a positive relationship: as advertising spending increases, sales generally increase.

## 5. Pie Chart

### Purpose
A pie chart shows how categories contribute to a whole.

### When to use
Use it when you want to show percentages or proportions of a total.

### Advantages
- Simple for small numbers of categories
- Clearly shows proportions
- Easy for basic business presentations

### Limitations
- Difficult to compare similar-sized slices
- Not suitable for many categories
- Less precise than a bar chart

```python
products = ["Laptop", "Phone", "Tablet", "Watch"]
sales = [120, 200, 150, 100]

plt.pie(
    sales,
    labels=products,
    autopct="%1.1f%%"
)

plt.title("Sales Distribution")
plt.show()
```

### Interpretation

The Phone category represents the largest share of total sales.

## 6. Density Plot

### Purpose
A density plot shows the smooth distribution of numerical data.

### When to use
Use it to understand the shape, concentration, and spread of data.

### Advantages
- Shows a smooth distribution
- Useful for comparing distributions
- Helps identify peaks and skewness

### Limitations
- Smoothing can hide some details
- Interpretation can be difficult for beginners

```python
data = np.random.normal(50, 10, 1000)

sns.kdeplot(data=data)

plt.xlabel("Value")
plt.ylabel("Density")
plt.title("Density Plot")
plt.show()
```

### Interpretation

The highest part of the curve represents where the data is most concentrated.

## 7. Box Plot

### Purpose
A box plot summarizes the distribution using the median, quartiles, and possible outliers.

### When to use
Use it to compare distributions and detect outliers.

### Advantages
- Quickly shows spread
- Shows median and quartiles
- Helps detect outliers

### Limitations
- Does not show every individual observation
- Can be difficult for beginners to interpret

```python
data = [10, 12, 13, 15, 16, 18, 20, 22, 25, 50]

plt.boxplot(data)
plt.ylabel("Values")
plt.title("Box Plot")
plt.show()
```

### Interpretation

The box shows the middle 50% of the data.

A point far away from the main box may represent an outlier.

## 8. Violin Plot

### Purpose
A violin plot combines information about data distribution with a box-plot-like summary.

### When to use
Use it to compare the distribution and density of multiple groups.

### Advantages
- Shows distribution shape
- Shows density
- Useful for comparing groups

### Limitations
- More difficult to interpret than a box plot
- Can become confusing with many groups

```python
group_a = [10, 12, 13, 14, 15, 16, 18]
group_b = [20, 21, 22, 24, 25, 27, 30]

plt.violinplot([group_a, group_b])

plt.xticks([1, 2], ["Group A", "Group B"])
plt.ylabel("Values")
plt.title("Violin Plot")
plt.show()
```

### Interpretation

The width of the violin shows where observations are more concentrated.

The two groups have different distributions and value ranges.

## 9. Pair Plot

### Purpose
A pair plot shows relationships between multiple numerical variables at the same time.

### When to use
Use it during Exploratory Data Analysis (EDA) to identify correlations, distributions, and patterns.

### Advantages
- Shows many relationships together
- Useful for EDA
- Helps identify possible correlations

### Limitations
- Can become crowded with many features
- Requires more computation for large datasets

```python
data = pd.DataFrame({
    "Age": [20, 25, 30, 35, 40],
    "Income": [20, 30, 40, 50, 60],
    "Spending": [15, 25, 35, 45, 50]
})

sns.pairplot(data)

plt.show()
```

### Interpretation

The pair plot allows us to examine relationships between Age, Income, and Spending simultaneously.

It can help identify positive or negative relationships between features.

## 10. Heatmap

### Purpose
A heatmap represents numerical values using different shades of color.

### When to use
Use it for correlation matrices, missing values, and other matrix-based data.

### Advantages
- Easy to identify patterns
- Useful for correlation analysis
- Can display many values compactly

### Limitations
- Too many variables can make it crowded
- Color differences may be difficult to interpret

```python
data = pd.DataFrame({
    "Age": [20, 25, 30, 35, 40],
    "Income": [20, 30, 40, 50, 60],
    "Spending": [15, 25, 35, 45, 50]
})

correlation = data.corr()

sns.heatmap(
    correlation,
    annot=True,
    cmap="coolwarm"
)

plt.title("Correlation Heatmap")
plt.show()
```

### Interpretation

The heatmap makes strong and weak correlations easier to identify.

Values close to +1 indicate strong positive correlation, while values close to -1 indicate strong negative correlation.
