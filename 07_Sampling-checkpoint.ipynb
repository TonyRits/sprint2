# Sampling

Sampling is the process of selecting a smaller group from a larger population to study the data.

In this notebook, we will learn different sampling methods, sampling bias, sampling error, and sample size.

## 1. Sampling

### Step 1 – Understand the Concept

Sampling means selecting a smaller group from a large population.

It helps us study data without collecting information from every member of the population.

### Step 2 – Demonstrate the Concept

Suppose a company has 10,000 customers.

Instead of asking all 10,000 customers, we can select 500 customers and analyze their feedback.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

population = np.arange(1, 10001)

sample = np.random.choice(population, size=500, replace=False)

print("Population Size:", len(population))
print("Sample Size:", len(sample))
print("First 10 Sample Values:", sample[:10])
```

### Interpretation

The population contains 10,000 customers, while the sample contains 500 customers.

Sampling reduces the time and cost required to collect data.

## 2. Random Sampling

### Step 1 – Understand the Concept

Random sampling gives every member of the population an equal chance of being selected.

The selection is done randomly without using specific groups or categories.

### Step 2 – Demonstrate the Concept

Suppose a college has 1,000 students.

We randomly select 100 students for a survey, giving every student an equal chance of selection.

```python
students = np.arange(1, 1001)

random_sample = np.random.choice(
    students,
    size=100,
    replace=False
)

print("Selected Students:", random_sample[:10])
print("Sample Size:", len(random_sample))
```

### Interpretation

The students were selected randomly from the complete population.

Random sampling helps reduce selection bias.

## 3. Stratified Sampling

### Step 1 – Understand the Concept

Stratified sampling divides the population into groups called strata.

Samples are then selected from each group.

### Step 2 – Demonstrate the Concept

Suppose a company has employees from three departments:

- AI
- Data
- HR

We can select employees from each department to make sure every department is represented.

```python
data = pd.DataFrame({
    "Name": ["A", "B", "C", "D", "E", "F", "G", "H"],
    "Department": ["AI", "AI", "Data", "Data", "HR", "HR", "AI", "Data"]
})

stratified_sample = (
    data.groupby("Department", group_keys=False)
    .apply(lambda x: x.sample(n=1), include_groups=False)
)

print(stratified_sample)
```

### Interpretation

A sample is selected from each department.

This ensures that important groups are represented in the sample.

## 4. Cluster Sampling

### Step 1 – Understand the Concept

Cluster sampling divides the population into groups called clusters.

Instead of selecting individuals from every cluster, we randomly select some complete clusters.

### Step 2 – Demonstrate the Concept

Suppose a company has 10 branches.

Instead of selecting employees from every branch, we randomly select 3 branches and survey employees from those branches.

```python
branches = ["Chennai", "Bangalore", "Mumbai", "Delhi", "Pune"]

selected_branches = np.random.choice(
    branches,
    size=2,
    replace=False
)

print("Selected Clusters:", selected_branches)
```

### Interpretation

Two branches were randomly selected as clusters.

All or many members inside the selected clusters can then be studied.

## 5. Systematic Sampling

### Step 1 – Understand the Concept

Systematic sampling selects members at a fixed interval from an ordered population.

For example, we can select every 10th person from a list.

### Step 2 – Demonstrate the Concept

Suppose we have 1,000 customers and want a sample of 100.

We can select every 10th customer after choosing a starting point.

```python
population = np.arange(1, 1001)

step = 10
start = 5

systematic_sample = population[start - 1::step]

print("First 10 Selected Values:", systematic_sample[:10])
print("Sample Size:", len(systematic_sample))
```

### Interpretation

Every 10th member is selected from the population.

Systematic sampling is simple and efficient for ordered data.

## 6. Sampling Bias

### Step 1 – Understand the Concept

Sampling bias occurs when some members of the population are more likely to be selected than others.
m
This can make the sample unrepresentative of the population.

### Step 2 – Demonstrate the Concept

Suppose a company asks only morning-shift employees about working conditions.

The results may not represent evening-shift employees.

```python
population = pd.DataFrame({
    "Shift": ["Morning"] * 50 + ["Evening"] * 50
})

biased_sample = population[
    population["Shift"] == "Morning"
].sample(20)

print(biased_sample["Shift"].value_counts())
```

### Interpretation

The sample contains only morning-shift employees.

Therefore, the sample may not represent the complete population.

## 7. Sampling Error

### Step 1 – Understand the Concept

Sampling error is the difference between a population value and the value calculated from a sample.

It occurs because we study only part of the population.

### Step 2 – Demonstrate the Concept

Suppose the actual population average is 50.

If a sample gives an average of 48, the difference is sampling error.

```python
population = np.arange(1, 101)

population_mean = np.mean(population)

sample = np.random.choice(
    population,
    size=20,
    replace=False
)

sample_mean = np.mean(sample)

sampling_error = population_mean - sample_mean

print("Population Mean:", population_mean)
print("Sample Mean:", sample_mean)
print("Sampling Error:", sampling_error)
```

### Interpretation

The sample mean may be different from the population mean.

The difference between them represents sampling error.

## 8. Sample Size

### Step 1 – Understand the Concept

Sample size is the number of observations selected from a population.

A suitable sample size helps produce reliable statistical results.

### Step 2 – Demonstrate the Concept

A very small sample may not represent the population well.

A larger and appropriate sample can provide more reliable estimates.

```python
population = np.arange(1, 10001)

small_sample = np.random.choice(
    population,
    size=50,
    replace=False
)

large_sample = np.random.choice(
    population,
    size=500,
    replace=False
)

print("Small Sample Size:", len(small_sample))
print("Large Sample Size:", len(large_sample))
```

### Interpretation

The first sample contains 50 observations and the second contains 500 observations.

The required sample size depends on the population, variability, confidence level, and required accuracy.
