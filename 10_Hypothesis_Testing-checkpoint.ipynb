# Hypothesis Testing

Hypothesis testing is a statistical method used to make decisions about a population using sample data.

In this notebook, we will learn null hypothesis, alternative hypothesis, p-value, confidence interval, significance level, errors, and types of tests.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from scipy import stats
```

## 1. Null Hypothesis

### Step 1 – Understand the Concept

The Null Hypothesis (H₀) states that there is no significant difference or effect.

It represents the default assumption that we test using sample data.

### Step 2 – Demonstrate the Concept

Suppose a company claims that the average delivery time is 30 minutes.

The null hypothesis can be:

H₀: Mean delivery time = 30 minutes

```python
sample = [28, 31, 30, 29, 32, 30, 31, 28, 30, 29]

sample_mean = np.mean(sample)

print("Sample Mean:", sample_mean)
```

### Interpretation

We compare the sample evidence with the assumption made in the null hypothesis.

If the evidence is strong enough, we may reject H₀.

## 2. Alternative Hypothesis

### Step 1 – Understand the Concept

The Alternative Hypothesis (H₁ or Ha) states that there is a significant difference or effect.

It represents the possibility that the null hypothesis is not true.

### Step 2 – Demonstrate the Concept

If the company claims that the average delivery time is 30 minutes:

H₀: Mean delivery time = 30 minutes

H₁: Mean delivery time ≠ 30 minutes

```python
hypothesized_mean = 30
sample_mean = np.mean(sample)

difference = sample_mean - hypothesized_mean

print("Sample Mean:", sample_mean)
print("Difference from Hypothesized Mean:", difference)
```

### Interpretation

The alternative hypothesis represents a difference or effect that we are testing for.

The statistical test helps us determine whether there is enough evidence against H₀.

## 3. p-value

### Step 1 – Understand the Concept

The p-value measures how compatible the observed data is with the null hypothesis.

A small p-value provides stronger evidence against the null hypothesis.

### Step 2 – Demonstrate the Concept

A common decision rule is:

If p-value < significance level → Reject H₀

If p-value ≥ significance level → Fail to reject H₀

```python
sample = [28, 31, 30, 29, 32, 30, 31, 28, 30, 29]

t_statistic, p_value = stats.ttest_1samp(
    sample,
    popmean=30
)

print("T-statistic:", t_statistic)
print("p-value:", p_value)
```

### Interpretation

The p-value is compared with the chosen significance level.

A p-value below the significance level provides evidence to reject H₀.

## 4. Confidence Interval

### Step 1 – Understand the Concept

A confidence interval gives a range of plausible values for a population parameter.

For example, a 95% confidence interval gives a range estimated using the sample data.

### Step 2 – Demonstrate the Concept

Suppose the average customer spending is estimated as ₹500.

A confidence interval might give a range such as ₹470 to ₹530.

```python
data = np.array([480, 510, 495, 520, 505, 490, 515, 500, 485, 510])

mean = np.mean(data)
std_error = stats.sem(data)

confidence_interval = stats.t.interval(
    confidence=0.95,
    df=len(data) - 1,
    loc=mean,
    scale=std_error
)

print("Mean:", mean)
print("95% Confidence Interval:", confidence_interval)
```

### Interpretation

The confidence interval provides a range of plausible values for the population mean.

A wider interval indicates more uncertainty, while a narrower interval indicates a more precise estimate.

## 5. Significance Level

### Step 1 – Understand the Concept

The significance level, represented by α (alpha), is the threshold used to decide whether to reject the null hypothesis.

A commonly used significance level is 0.05 or 5%.

### Step 2 – Demonstrate the Concept

If α = 0.05:

p-value < 0.05 → Reject H₀

p-value ≥ 0.05 → Fail to reject H₀

```python
alpha = 0.05

print("Significance Level:", alpha)

if p_value < alpha:
    print("Reject the Null Hypothesis")
else:
    print("Fail to Reject the Null Hypothesis")
```

### Interpretation

The significance level defines how much evidence is required to reject the null hypothesis.

A smaller alpha means a stricter threshold.

## 6. Type I Error

### Step 1 – Understand the Concept

Type I Error occurs when we reject a true null hypothesis.

It is also called a false positive.

### Step 2 – Demonstrate the Concept

Suppose a medical test concludes that a healthy person has a disease.

The test has incorrectly rejected the true condition of the null hypothesis.

### Interpretation

Type I Error means:

True H₀ → Rejected H₀

The probability of Type I Error is related to the significance level α.

## 7. Type II Error

### Step 1 – Understand the Concept

Type II Error occurs when we fail to reject a false null hypothesis.

It is also called a false negative.

### Step 2 – Demonstrate the Concept

Suppose a medical test fails to detect a disease in a person who actually has the disease.

This is an example of a false negative.

### Interpretation

Type II Error means:

False H₀ → Fail to Reject H₀

## 8. One-tailed Test

### Step 1 – Understand the Concept

A one-tailed test checks for an effect in one specific direction.

The alternative hypothesis uses either > or <.

### Step 2 – Demonstrate the Concept

Suppose a company wants to check whether a new process increases productivity.

H₀: Mean productivity ≤ 100

H₁: Mean productivity > 100

This is a right-tailed test.

```python
sample = np.array([105, 110, 108, 112, 107, 109, 111, 106])

t_statistic, two_tailed_p = stats.ttest_1samp(
    sample,
    popmean=100
)

one_tailed_p = two_tailed_p / 2

print("T-statistic:", t_statistic)
print("One-tailed p-value:", one_tailed_p)
```

### Interpretation

A one-tailed test is used when the research question is concerned with a specific direction.

For example, whether a value is greater than or less than a particular value.

## 9. Two-tailed Test

### Step 1 – Understand the Concept

A two-tailed test checks whether a value is significantly different in either direction.

The alternative hypothesis uses ≠.

### Step 2 – Demonstrate the Concept

Suppose a company wants to know whether the average delivery time is different from 30 minutes.

H₀: Mean delivery time = 30

H₁: Mean delivery time ≠ 30

```python
sample = np.array([28, 31, 30, 29, 32, 30, 31, 28])

t_statistic, p_value = stats.ttest_1samp(
    sample,
    popmean=30
)

print("T-statistic:", t_statistic)
print("Two-tailed p-value:", p_value)
```

### Interpretation

A two-tailed test checks for significant differences on both sides of the hypothesized value.

It is useful when we do not specify the direction of the difference in advance.
