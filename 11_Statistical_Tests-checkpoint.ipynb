# Statistical Tests

Statistical tests help us determine whether a difference or relationship in data is statistically significant.

They are commonly used in business analysis, experiments, and AI/ML.

```python
import numpy as np
import pandas as pd
from scipy import stats
```

## 1. T-Test

### Definition
A T-Test checks whether the means of one or two groups are significantly different.

It is mainly used when the sample size is small or the population standard deviation is unknown.

- Data should be approximately normally distributed.
- Observations should be independent.
- For an independent T-Test, groups should be independent.
- For equal-variance T-Test, group variances should be similar.

Use a T-Test when comparing the mean of one or two groups.

Do not use it for highly non-normal data or when observations are not independent.

A company wants to check whether the average sales before and after a marketing campaign are different.

```python
before = [50, 52, 48, 49, 51, 47, 53]
after = [55, 57, 54, 56, 58, 53, 59]

result = stats.ttest_rel(before, after)

print("T-statistic:", result.statistic)
print("P-value:", result.pvalue)
```

### Interpretation

If p-value < 0.05, we reject the null hypothesis and conclude that the two means are significantly different.

## 2. Z-Test

### Definition
A Z-Test checks whether a sample mean is significantly different from a known population mean.

It is generally used when the population standard deviation is known and the sample size is large.

- Population standard deviation is known.
- Observations are independent.
- Data is normally distributed or sample size is sufficiently large.

Use a Z-Test for large samples when population standard deviation is known.

Do not use it when population standard deviation is unknown and the sample is small.

A company knows the population standard deviation of delivery time and wants to check whether the average delivery time has changed.

```python
sample_mean = 52
population_mean = 50
population_std = 5
sample_size = 100

z_score = (sample_mean - population_mean) / (population_std / np.sqrt(sample_size))

p_value = 2 * (1 - stats.norm.cdf(abs(z_score)))

print("Z-score:", z_score)
print("P-value:", p_value)
```

### Interpretation

If p-value < 0.05, the sample mean is significantly different from the population mean.

## 3. Chi-Square Test

### Definition
A Chi-Square Test checks whether categorical variables are associated or whether observed frequencies differ from expected frequencies.

- Data should be categorical.
- Observations should be independent.
- Expected frequencies should generally be sufficiently large.

Use it for categorical data and frequency/count comparisons.

Do not use it directly for continuous numerical data.

A company wants to know whether customer gender and product preference are related.

```python
data = np.array([
    [40, 30],
    [20, 50]
])

chi2, p_value, dof, expected = stats.chi2_contingency(data)

print("Chi-square:", chi2)
print("P-value:", p_value)
print("Expected frequencies:")
print(expected)
```

### Interpretation

If p-value < 0.05, there is a statistically significant association between the categorical variables.

## 4. ANOVA

### Definition
ANOVA (Analysis of Variance) tests whether the means of three or more groups are significantly different.

It compares variation between groups with variation within groups.

- Observations should be independent.
- Data should be approximately normally distributed.
- Groups should have similar variances.

Use ANOVA when comparing the means of three or more groups.

Do not use it when observations are dependent or assumptions are strongly violated.

A company wants to compare the average sales generated by three different marketing strategies.

```python
strategy_A = [20, 22, 21, 23, 19]
strategy_B = [25, 27, 26, 28, 24]
strategy_C = [30, 32, 31, 33, 29]

f_stat, p_value = stats.f_oneway(
    strategy_A,
    strategy_B,
    strategy_C
)

print("F-statistic:", f_stat)
print("P-value:", p_value)
```

### Interpretation

If p-value < 0.05, at least one group mean is significantly different from the others.

## 5. Mann-Whitney U Test

### Definition
The Mann-Whitney U Test is a non-parametric test used to compare two independent groups.

It is an alternative to the independent T-Test when normality assumptions are not satisfied.

- Two groups should be independent.
- Data should be at least ordinal.
- Observations should be independent.

Use it when comparing two independent groups with non-normal or ordinal data.

Do not use it for paired/dependent observations.

A company wants to compare customer satisfaction scores between two branches.

```python
branch_A = [70, 75, 80, 65, 72]
branch_B = [85, 90, 88, 92, 87]

result = stats.mannwhitneyu(
    branch_A,
    branch_B,
    alternative="two-sided"
)

print("U-statistic:", result.statistic)
print("P-value:", result.pvalue)
```

### Interpretation

If p-value < 0.05, the two groups have a statistically significant difference in their distributions.

## 6. Shapiro-Wilk Test

### Definition
The Shapiro-Wilk Test checks whether a dataset is approximately normally distributed.

- Observations should be independent.
- Data should be continuous or approximately continuous.
- The test is mainly suitable for small or moderate sample sizes.

Use it to check normality before applying tests that assume normally distributed data.

Do not rely only on this test for very large datasets because small deviations can become statistically significant.

Do not rely only on this test for very large datasets because small deviations can become statistically significant.

```python
data = [48, 51, 49, 52, 50, 47, 53, 51, 49, 50]

statistic, p_value = stats.shapiro(data)

print("Shapiro-Wilk statistic:", statistic)
print("P-value:", p_value)
```

### Interpretation

If p-value > 0.05, we do not reject the null hypothesis, so the data is considered approximately normal.

If p-value < 0.05, the data shows evidence of non-normality.
