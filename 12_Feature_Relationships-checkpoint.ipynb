# Feature Relationships

Feature relationships help us understand how input variables are related to each other and how useful they may be for Machine Learning models.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.datasets import load_diabetes
from sklearn.ensemble import RandomForestRegressor
from statsmodels.stats.outliers_influence import variance_inflation_factor
```

## 1. Correlation Matrix

### Definition

A correlation matrix shows the correlation between multiple numerical variables.

The value ranges from -1 to +1.

Use it to quickly identify positive, negative, or weak relationships between features.

A company can check whether advertising spend, sales, profit, and customer visits are related.

```python
data = pd.DataFrame({
    "Advertising": [10, 20, 30, 40, 50],
    "Sales": [15, 25, 35, 45, 55],
    "Customers": [100, 120, 140, 160, 180],
    "Profit": [5, 10, 15, 20, 25]
})

correlation_matrix = data.corr()

print(correlation_matrix)
```

```python
plt.imshow(correlation_matrix, cmap="coolwarm")
plt.colorbar()

plt.xticks(range(len(correlation_matrix.columns)), correlation_matrix.columns, rotation=45)
plt.yticks(range(len(correlation_matrix.columns)), correlation_matrix.columns)

plt.title("Correlation Matrix")
plt.show()
```

### Interpretation

Values close to +1 indicate strong positive correlation.

Values close to -1 indicate strong negative correlation, while values close to 0 indicate weak or no linear correlation.

## 2. Multicollinearity

### Definition

Multicollinearity occurs when two or more independent features are highly correlated with each other.

This can make some Machine Learning models difficult to interpret.

Check for multicollinearity when building models such as Linear Regression or Logistic Regression.

It is less concerning for tree-based models such as Random Forest, although redundant features can still exist.

A company uses both "monthly income" and "annual income" as features.
They contain almost the same information and may create multicollinearity.

```python
data = pd.DataFrame({
    "Monthly_Income": [20, 30, 40, 50, 60],
    "Annual_Income": [240, 360, 480, 600, 720],
    "Age": [22, 30, 35, 42, 50]
})

print(data.corr())
```

### Interpretation

If two independent features have a very high correlation, they may have multicollinearity.

Removing one redundant feature can sometimes improve model stability and interpretation.

## 3. Variance Inflation Factor (VIF)

### Definition

VIF measures how much the variance of a feature's coefficient is increased because of relationships with other features.

Higher VIF indicates stronger multicollinearity.

VIF = 1 / (1 - R²)

- VIF = 1 → No multicollinearity
- VIF between 1 and 5 → Usually acceptable
- VIF > 5 → Possible multicollinearity
- VIF > 10 → Strong multicollinearity

A company uses advertising cost, marketing budget, and campaign spending as model features.

VIF can identify whether these features contain highly overlapping information.

```python
data = pd.DataFrame({
    "Advertising": [10, 20, 30, 40, 50, 60],
    "Marketing": [12, 24, 36, 48, 60, 72],
    "Customers": [100, 110, 125, 140, 155, 170]
})

vif_data = pd.DataFrame()
vif_data["Feature"] = data.columns

vif_data["VIF"] = [
    variance_inflation_factor(data.values, i)
    for i in range(data.shape[1])
]

print(vif_data)
```

### Interpretation

A high VIF means that a feature can be strongly explained by other features.

For regression models, features with very high VIF may need to be removed or transformed.

## 4. Feature Importance

### Definition

Feature importance shows how much each feature contributes to a Machine Learning model's prediction.

Different models calculate feature importance differently.

Use feature importance to understand which features contribute most to predictions.

A sales prediction model can identify whether price, advertising, quantity, or customer age contributes more to predicted sales.

```python
from sklearn.datasets import load_diabetes
from sklearn.ensemble import RandomForestRegressor

diabetes = load_diabetes()

X = pd.DataFrame(
    diabetes.data,
    columns=diabetes.feature_names
)

y = diabetes.target

model = RandomForestRegressor(
    n_estimators=100,
    random_state=42
)

model.fit(X, y)

importance = pd.DataFrame({
    "Feature": X.columns,
    "Importance": model.feature_importances_
})

importance = importance.sort_values(
    by="Importance",
    ascending=False
)

print(importance)
```

```python
plt.bar(importance["Feature"], importance["Importance"])

plt.xlabel("Features")
plt.ylabel("Importance")
plt.title("Feature Importance")

plt.xticks(rotation=45)
plt.show()
```

### Interpretation

A higher importance value means the feature contributed more to the Random Forest model's predictions.

Feature importance does not always mean that the feature causes the target variable to change.

## 5. Feature Selection Basics

### Definition

Feature selection means selecting the most useful features and removing unnecessary or redundant features.

It can make models simpler, faster, and sometimes more accurate.

Common approaches include:

- Correlation-based selection
- Statistical tests
- Feature importance
- Recursive Feature Elimination
- VIF-based selection

Use feature selection when the dataset contains many features, irrelevant features, or highly correlated features.

Do not remove a feature only because its individual correlation is low; some features may become useful when combined with others.

A customer churn model may contain 50 features.

Feature selection can remove irrelevant features and keep the variables that provide useful information for predicting churn.

```python
from sklearn.feature_selection import SelectKBest, f_regression

diabetes = load_diabetes()

X = pd.DataFrame(
    diabetes.data,
    columns=diabetes.feature_names
)

y = diabetes.target

selector = SelectKBest(
    score_func=f_regression,
    k=5
)

X_selected = selector.fit_transform(X, y)

selected_features = X.columns[selector.get_support()]

print("Selected Features:")
print(selected_features.tolist())
```

### Interpretation

SelectKBest selects the top K features based on a statistical score.

Here, the top 5 features are selected based on their relationship with the target variable.
