## Question 1: Explain the difference between Mean and Median

### Answer

Mean is the average of all values in a dataset.

Median is the middle value after arranging the data in ascending or descending order.

Median is less affected by extreme values or outliers than the mean.

### Example

Data = [10, 20, 30, 40, 100]

Mean = 40

Median = 30

The outlier 100 increases the mean, while the median remains closer to the typical values.

## Question 2: What is Variance?

Variance measures how much the values in a dataset are spread around the mean.

It is calculated using the average of the squared differences from the mean.

### Formula

Population Variance:

Variance = Σ(x - μ)² / N

Variance helps us understand the amount of variability in the data.

## Question 3: What is Standard Deviation?

Standard deviation measures the typical distance of data values from the mean.

It is the square root of variance and is expressed in the same units as the original data.

### Formula

Standard Deviation = √Variance

A small standard deviation means the values are close to the mean, while a large standard deviation means the values are more spread out.

## Question 4: Explain Bayes' Theorem with an Example

Bayes' Theorem calculates the probability of an event based on new evidence.

It is commonly used in classification and medical diagnosis.

### Formula

P(A|B) = [P(B|A) × P(A)] / P(B)

Where:

- P(A|B) = Probability of A given B
- P(B|A) = Probability of B given A
- P(A) = Prior probability of A
- P(B) = Probability of evidence B

### Example

Suppose 20% of emails are spam.

If an email is spam, there is an 80% chance that it contains a certain suspicious word.

Using Bayes' Theorem, we can calculate the probability that an email is spam when that word is found.

```python
P_spam = 0.20
P_word_given_spam = 0.80
P_word = 0.30

P_spam_given_word = (P_word_given_spam * P_spam) / P_word

print("Probability of Spam given the word:", P_spam_given_word)
```

### Interpretation

The result represents the probability that an email is spam when the suspicious word is present.

## Question 6: Explain Correlation and Covariance

Covariance shows the direction in which two variables move together.

Correlation measures both the direction and strength of the relationship and ranges from -1 to +1.

### Difference

- Positive covariance/correlation: variables tend to increase together.
- Negative covariance/correlation: one tends to increase when the other decreases.
- Correlation has a fixed range from -1 to +1.
- Covariance does not have a fixed range.

## Question 8: Explain IQR

IQR stands for Interquartile Range.

It represents the middle 50% of the data.

Formula:

IQR = Q3 - Q1

### Example

If Q1 = 20 and Q3 = 40:

IQR = 40 - 20 = 20

IQR is commonly used to detect outliers.

## Question 9: What is a p-value?

A p-value indicates how compatible the observed result is with the null hypothesis.

A commonly used significance level is 0.05.

If p-value < 0.05, we reject the null hypothesis.
If p-value >= 0.05, we fail to reject the null hypothesis.

## Question 12: Detect Outliers using the IQR Method

The IQR method identifies values below the lower bound or above the upper bound.

Lower Bound = Q1 - 1.5 × IQR

Upper Bound = Q3 + 1.5 × IQR

```python
data = np.array([10, 12, 11, 13, 12, 50, 14, 15])

Q1 = np.percentile(data, 25)
Q3 = np.percentile(data, 75)

IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

outliers = data[(data < lower_bound) | (data > upper_bound)]

print("Q1:", Q1)
print("Q3:", Q3)
print("IQR:", IQR)
print("Lower Bound:", lower_bound)
print("Upper Bound:", upper_bound)
print("Outliers:", outliers)
```

### Interpretation

The value 50 is detected as an outlier because it falls outside the calculated lower and upper bounds.

## Question 13: Calculate Correlation between Two Variables

Correlation measures the strength and direction of the relationship between two numerical variables.

A value close to +1 indicates strong positive correlation, while a value close to -1 indicates strong negative correlation.

```python
x = [10, 20, 30, 40, 50]
y = [15, 25, 35, 45, 55]

correlation = np.corrcoef(x, y)[0, 1]

print("Correlation:", correlation)
```

### Interpretation

The correlation is 1.0, which indicates a perfect positive correlation between the two variables.
