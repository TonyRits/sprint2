# Probability Distributions

Probability distributions describe how probabilities are distributed across possible values of a random variable.

In this notebook, we will learn PMF, PDF, CDF, Bernoulli, Binomial, Poisson, Normal, and Uniform distributions.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from scipy.stats import bernoulli, binom, poisson, norm, uniform
```

## 1. Bernoulli Distribution

### Step 1 – Understand the Concept

Bernoulli distribution represents an experiment with only two possible outcomes.

The outcomes are usually represented as 0 and 1.

### Step 2 – Demonstrate the Concept

A single coin toss can be represented as:

0 = Tail
1 = Head

If the coin is fair, both outcomes have probability 0.5.

```python
p = 0.5

x = [0, 1]
pmf = bernoulli.pmf(x, p)

print("P(0):", pmf[0])
print("P(1):", pmf[1])
```

### Interpretation

Both outcomes have a probability of 0.5.

Bernoulli distribution is useful for binary classification problems.

## 2. Binomial Distribution

### Step 1 – Understand the Concept

Binomial distribution represents the number of successes in a fixed number of independent trials.

Each trial has two possible outcomes.

### Step 2 – Demonstrate the Concept

Suppose a coin is tossed 10 times.

The Binomial distribution can tell us the probability of getting exactly 6 Heads.

```python
n = 10
p = 0.5
k = 6

probability = binom.pmf(k, n, p)

print("Probability of exactly 6 Heads:", probability)
```

### Interpretation

The result represents the probability of getting exactly 6 Heads in 10 tosses.

Binomial distribution is useful for binary outcomes and success/failure experiments.

## 3. Poisson Distribution

### Step 1 – Understand the Concept

Poisson distribution models the number of times an event occurs within a fixed interval.

The interval can represent time, distance, or area.

### Step 2 – Demonstrate the Concept

Suppose a customer support team receives an average of 3 calls per hour.

Poisson distribution can estimate the probability of receiving exactly 5 calls in an hour.

```python
lambda_value = 3
k = 5

probability = poisson.pmf(k, lambda_value)

print("Probability of exactly 5 calls:", probability)
```

### Interpretation

The result gives the probability of receiving exactly 5 calls in one hour.

Poisson distribution is useful for count-based events.

## 4. Uniform Distribution

### Step 1 – Understand the Concept

In a uniform distribution, values within a specified range have equal probability density.


### Step 2 – Demonstrate the Concept

A random number generated between 0 and 1 follows a uniform distribution when every value in that range has equal density.

```python
data = np.random.uniform(0, 1, 1000)

plt.hist(data, bins=20)
plt.xlabel("Value")
plt.ylabel("Frequency")
plt.title("Uniform Distribution")
plt.show()
```

### Interpretation

The values are spread approximately evenly between 0 and 1.

Uniform distribution is useful when all values in a range are treated equally.

## 5. Normal Distribution

### Step 1 – Understand the Concept

Normal distribution is a symmetric, bell-shaped distribution.

The mean, median, and mode are equal in a perfectly normal distribution.

### Step 2 – Demonstrate the Concept

Examples such as measurement errors or some naturally varying measurements can sometimes be approximately normally distributed.

The mean represents the center of the distribution.

```python
data = np.random.normal(50, 10, 1000)

plt.hist(data, bins=30)
plt.xlabel("Value")
plt.ylabel("Frequency")
plt.title("Normal Distribution")
plt.show()
```

### Interpretation

The values form an approximately bell-shaped pattern around the mean.

Normal distribution is widely used in statistics and machine learning.

```python
normal_data = np.random.normal(0, 1, 1000)
uniform_data = np.random.uniform(0, 1, 1000)

plt.hist(normal_data, bins=30, alpha=0.6, label="Normal")
plt.hist(uniform_data, bins=30, alpha=0.6, label="Uniform")

plt.xlabel("Value")
plt.ylabel("Frequency")
plt.title("Normal vs Uniform Distribution")
plt.legend()
plt.show()
```

### Interpretation

Normal distribution has a bell-shaped pattern.

Uniform distribution spreads values more evenly across its range.

## 6. Exponential Distribution

### Step 1 – Understand the Concept

Exponential distribution models the waiting time until an event occurs.

It is commonly used for time between events.

### Step 2 – Demonstrate the Concept

Suppose customers arrive at a service counter at a particular average rate.

Exponential distribution can be used to model the waiting time until the next customer arrives.

```python
data = np.random.exponential(scale=2, size=1000)

plt.hist(data, bins=30)
plt.xlabel("Waiting Time")
plt.ylabel("Frequency")
plt.title("Exponential Distribution")
plt.show()
```

### Interpretation

The distribution has many smaller waiting times and fewer larger waiting times.

Exponential distribution is useful for modeling waiting times.
