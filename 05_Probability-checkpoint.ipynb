# Probability

Probability is the study of how likely an event is to happen.

In this notebook, we will learn about events, sample space, independent and dependent events, conditional probability, Bayes' theorem, probability rules, random variables, and expected value.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

## 1. Introduction to Probability

### Step 1 – Understand the Concept

Probability measures how likely an event is to happen.

Its value ranges from 0 to 1:
- 0 means impossible
- 1 means certain

Formula:

P(A) = Favorable Outcomes / Total Outcomes

### Step 2 – Demonstrate the Concept

For a fair coin, there are two possible outcomes: Head and Tail.

Probability of getting Head = 1/2 = 0.5.

```python
favorable_outcomes = 1
total_outcomes = 2

probability = favorable_outcomes / total_outcomes

print("Probability of getting Head:", probability)
```

### Interpretation

The probability of getting Head is 0.5 or 50%.

In AI/ML, probability is used to represent uncertainty and prediction confidence.

## 2. Events

### Step 1 – Understand the Concept

An event is an outcome or group of outcomes that we are interested in.

For example, getting an even number when rolling a die is an event.

### Step 2 – Demonstrate the Concept

For a die, the possible outcomes are 1, 2, 3, 4, 5, and 6.

The event of getting an even number is {2, 4, 6}.

```python
outcomes = {1, 2, 3, 4, 5, 6}
event_even = {2, 4, 6}

probability = len(event_even) / len(outcomes)

print("Probability of an even number:", probability)
```

### Interpretation

There are 3 favorable outcomes out of 6.

Therefore, the probability of getting an even number is 0.5 or 50%.

## 3. Sample Space

### Step 1 – Understand the Concept

A sample space is the complete set of all possible outcomes of an experiment.

It is usually represented by S.

### Step 2 – Demonstrate the Concept

When rolling a die, the sample space contains six possible outcomes.

S = {1, 2, 3, 4, 5, 6}

```python
sample_space = {1, 2, 3, 4, 5, 6}

print("Sample Space:", sample_space)
print("Number of outcomes:", len(sample_space))
```

### Interpretation

The sample space contains 6 possible outcomes.

Probability calculations are based on the possible outcomes in the sample space.

## 4. Independent Events

### Step 1 – Understand the Concept

Two events are independent when one event does not affect the probability of the other.

Formula:

P(A ∩ B) = P(A) × P(B)

### Step 2 – Demonstrate the Concept

Getting Head on one coin toss does not affect the next coin toss.

Therefore, the two coin tosses are independent.

## 5. Dependent Events

### Step 1 – Understand the Concept

Dependent events are events where one event affects the probability of another event.

Formula:

P(A ∩ B) = P(A) × P(B|A)

### Step 2 – Demonstrate the Concept

Suppose a bag contains 3 red and 2 blue balls.

If one red ball is removed, the probability of getting another red ball changes.

```python
red = 3
blue = 2
total = red + blue

p_first_red = red / total
p_second_red = (red - 1) / (total - 1)

joint_probability = p_first_red * p_second_red

print("Probability of two red balls:", joint_probability)
```

### Interpretation

The probability changes after the first ball is removed.

Therefore, these are dependent events.

## 6. Conditional Probability

### Step 1 – Understand the Concept

Conditional probability is the probability of an event occurring when another event is already known to have happened.

Formula:

P(A|B) = P(A ∩ B) / P(B)

```python
p_premium_and_app = 0.30
p_app = 0.60

conditional_probability = p_premium_and_app / p_app

print("P(Premium | App User):", conditional_probability)
```

### Interpretation

The result is 0.5 or 50%.

This means that, in this example, 50% of app users are premium customers.

Conditional probability is commonly used in classification and Bayesian models.

## 7. Bayes' Theorem

### Step 1 – Understand the Concept

Bayes' theorem updates the probability of an event using new evidence.

Formula:

P(A|B) = [P(B|A) × P(A)] / P(B)

### Step 2 – Demonstrate the Concept

Suppose an email has a 40% probability of being spam.

If certain words appear in the email, Bayes' theorem can update the probability that the email is spam.

```python
p_spam = 0.40
p_words_given_spam = 0.80
p_words = 0.50

p_spam_given_words = (p_words_given_spam * p_spam) / p_words

print("Probability of Spam given words:", p_spam_given_words)
```

### Interpretation

The updated probability is 0.64 or 64%.

Bayes' theorem is used in probabilistic machine learning and classification.

## 8. Probability Rules

### Step 1 – Understand the Concept

Probability rules help calculate probabilities of individual and combined events.

Important rules include complement, addition, and multiplication.

### Step 2 – Demonstrate the Concept

For a fair die, the probability of getting an even number is 3/6.

The probability of not getting an even number is:

1 - 3/6 = 0.5

```python
p_even = 3 / 6

p_not_even = 1 - p_even

print("P(Even):", p_even)
print("P(Not Even):", p_not_even)
```

### Important Formulas

Complement:

P(Aᶜ) = 1 - P(A)

Addition:

P(A ∪ B) = P(A) + P(B) - P(A ∩ B)

Multiplication:

P(A ∩ B) = P(A) × P(B|A)

## 9. Random Variables

### Step 1 – Understand the Concept

A random variable assigns a numerical value to the outcome of a random experiment.

Random variables can be discrete or continuous.

### Step 2 – Demonstrate the Concept

For a dice experiment, the random variable X can represent the number obtained.

X can have values from 1 to 6.

```python
rolls = np.random.randint(1, 7, size=10)

print("Random variable values:", rolls)
```

### Interpretation

Each generated number represents one possible value of the random variable.

Random variables are important when modeling uncertain data.

## 10. Expected Value

### Step 1 – Understand the Concept

Expected value represents the average outcome we expect over many repeated experiments.

Formula:

E(X) = Σ xP(x)

### Step 2 – Demonstrate the Concept

For a fair six-sided die, every number has probability 1/6.

The expected value is 3.5.

```python
values = np.array([1, 2, 3, 4, 5, 6])
probabilities = np.array([1/6] * 6)

expected_value = np.sum(values * probabilities)

print("Expected Value:", expected_value)
```

### Interpretation

The expected value is 3.5.

It does not mean that we can roll a 3.5. It represents the long-term average outcome.
